**What is TextBlob?**    

TextBlob objects can be thought of as Python strings that learned how to do NLP.

## 1. Create a TextBlob

```python
from textblob import TextBlob
```

```python
wiki = TextBlob("Python is a high-level, general-purpose programming language.")
```

This makes a textblob object called wiki.

## 2. Part-of-speech Tagging

```python
wiki.tags
```

Each of these tags correspond to a part of speech. For example, NNP means proper oun (singular). VBZ means verb in present tense, and so on.

## 3. Noun Phrase Extraction

```python
wiki.noun_phrases
```

Returns all noun phrases as a wordlist.

## 4. Sentiment Analysis

Returns a named tuple of the form Sentiment (polarity, subjectivity). 

```python
testimonial = TextBlob("Textblob is amazingly simple to use. What great fun!")
```

```python
testimonial.sentiment
```

- Polarity (-1 to 1): overall tone of an article, how positive/negative/neutral it is.
- Subjectivity (0 to 1): 0 = objective and 1 = subjective.

## 5. Tokenization

```python
zen = TextBlob(
    "Beautiful is better than ugly. "
    "Explicit is better than implicit. "
    "Simple is better than complex."
)
zen.words
```

Each word is now its own token.

```python
zen.sentences
```

Or, each sentence could be its own token. 

```python
for sentence in zen.sentences:
    print(sentence.sentiment)
```

Each sentence object has the same properties and methods as TextBlobs.

## 6. Words Inflection and Lemmatization

```python
sentence = TextBlob("Use 4 spaces per indentation level.")
sentence.words
```

Each Word object has useful methods.

```python
sentence.words[2].singularize()
```

Makes the word into a singular form.

```python
sentence.words[-1].pluralize()
```

Makes the word into plural form.

```python
from textblob import Word
w = Word("octopi")
w.lemmatize()
```

Lemmatization: depluralise.

```python
w = Word('went')
w.lemmatize('v')
```

## 7. WordNet Integration

```python
from textblob import Word
from textblob.wordnet import VERB
word = Word("octopus")
word.synsets
```

Synsets: synonym sets, groups of interchangeable words that share the same distinct meaning or concept within a lexical database.

```python
Word("hack").get_synsets(pos=VERB)
```

```python
Word("octopus").definitions
```

Or could create synsets directly:

```python
from textblob.wordnet import Synset
octopus = Synset("octopus.n.02")
shrimp = Synset("shrimp.n.03")
octopus.path_similarity(shrimp)
```

## 8. WordLists

```python
animals = TextBlob("cat dog octopus")
animals.words
```

```python
animals.words.pluralize()
```

A python list with additional methods that can act on each word.

## 9. Spelling Correction

```python
b = TextBlob("I havv goood speling!")
print(b.correct())
```

```python
w = Word("falibility")
w.spellcheck()
```

Returns a list of (word, confidence) tuples with spelling suggestions.

Apparently, the spelling correction is only about 70% accurate, so we should use with precaution.

## 10. Get Word and Noun Phrase Frequencies

```python
monty = TextBlob("We are no longer the Knights who say Ni. "
                    "We are now the Knights who say Ekki ekki ekki PTANG.")
monty.word_counts['ekki']
```

THis means ekki appears 3 times.

```python
monty.words.count('ekki')
```

This method achieves the same effect.

```python
monty.words.count('ekki', case_sensitive=True)
```

This means we're only searching for ekki in the all lower-case form.

## 11. Parsing

```python
b = TextBlob("And now for something completely different.")
print(b.parse())
```

This means that the parser analyzes the sentence to break it down into grammatical components.

## 12. TextBlobs Are Like Python Strings!

Substringing (by character)

```python
zen[0:19]
```

Comparing between TextBlobs and strings.

```python
apple_blob = TextBlob("apples")
banana_blob = TextBlob("bananas")
apple_blob < banana_blob
```

lesser than or greater than in terms of lengths of words!

Concatenating and interpolating TextBlobs and strings.

```python
apple_blob + " and " + banana_blob
```

This makes a textblob object.

```python
"{0} and {1}".format(apple_blob, banana_blob)
```

returns a string.

## 13. n-grams

```python
blob = TextBlob("Now is better than never.")
blob.ngrams(n=3)
```

Returns a list of tuples of n successive words -> possibly useful for defining tokens of your chosen length.

## 14. Get Start and End Indices of Sentences

```python
for s in zen.sentences:
    print(s)
    print("---- Starts at index {}, Ends at index {}".format(s.start, s.end))
```

Where a sentence starts and ends within a textblob object.
